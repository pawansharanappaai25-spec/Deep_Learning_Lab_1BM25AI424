```python

# ============================================================
# SELF LEARNING LAB PROGRAM 3
# Plain CNN vs Residual CNN
# CIFAR-10 + MobileNetV2 + Grad-CAM
# ============================================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import time

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

SEED = 42
tf.keras.utils.set_random_seed(SEED)

BATCH_SIZE = 64
EPOCHS_CNN = 5
EPOCHS_MOBILENET = 5
NUM_CLASSES = 10

# Use a subset of CIFAR-10
TRAIN_SAMPLES = 5000
TEST_SAMPLES = 1000

# ------------------------------------------------------------
# 2. LOAD CIFAR-10 DATASET
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = cifar10.load_data()

x_train = x_train[:TRAIN_SAMPLES]
y_train = y_train[:TRAIN_SAMPLES]

x_test = x_test[:TEST_SAMPLES]
y_test = y_test[:TEST_SAMPLES]

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)

class_names = [
    'airplane', 'automobile', 'bird', 'cat', 'deer',
    'dog', 'frog', 'horse', 'ship', 'truck'
]

# ------------------------------------------------------------
# 3. DISPLAY SAMPLE IMAGES
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[int(y_train[i][0])])
    plt.axis('off')

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. DATA AUGMENTATION
# ------------------------------------------------------------

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])

# Normalize images to [0, 1]
x_train_norm = x_train.astype("float32") / 255.0
x_test_norm = x_test.astype("float32") / 255.0

# ------------------------------------------------------------
# 5. BUILD PLAIN CNN
# ------------------------------------------------------------

def build_plain_cnn():

    inputs = layers.Input(shape=(32, 32, 3))

    x = data_augmentation(inputs)

    x = layers.Conv2D(
        32, 3, padding='same', activation='relu'
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(
        64, 3, padding='same', activation='relu'
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(
        128, 3, padding='same', activation='relu'
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.GlobalAveragePooling2D()(x)

    x = layers.Dense(128, activation='relu')(x)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES, activation='softmax'
    )(x)

    model = models.Model(inputs, outputs)

    return model


plain_model = build_plain_cnn()

plain_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

plain_model.summary()

# ------------------------------------------------------------
# 6. BUILD RESIDUAL BLOCK
# ------------------------------------------------------------

def residual_block(x, filters, stride=1):

    shortcut = x

    # First convolution
    x = layers.Conv2D(
        filters, 3,
        strides=stride,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    # Second convolution
    x = layers.Conv2D(
        filters, 3,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)

    # Adjust shortcut dimensions if necessary
    if stride != 1 or shortcut.shape[-1] != filters:

        shortcut = layers.Conv2D(
            filters, 1,
            strides=stride,
            padding='same',
            use_bias=False
        )(shortcut)

        shortcut = layers.BatchNormalization()(shortcut)

    # Residual connection
    x = layers.Add()([x, shortcut])
    x = layers.ReLU()(x)

    return x


# ------------------------------------------------------------
# 7. BUILD RESIDUAL CNN
# ------------------------------------------------------------

def build_residual_cnn():

    inputs = layers.Input(shape=(32, 32, 3))

    x = data_augmentation(inputs)

    x = layers.Conv2D(
        32, 3, padding='same', use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    x = residual_block(x, 32)
    x = residual_block(x, 64, stride=2)
    x = residual_block(x, 128, stride=2)

    x = layers.GlobalAveragePooling2D()(x)

    x = layers.Dense(128, activation='relu')(x)
    x = layers.Dropout(0.3)(x)

    outputs = layers.Dense(
        NUM_CLASSES, activation='softmax'
    )(x)

    model = models.Model(inputs, outputs)

    return model


residual_model = build_residual_cnn()

residual_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

residual_model.summary()

# ------------------------------------------------------------
# 8. TRAIN PLAIN CNN
# ------------------------------------------------------------

print("\nTraining Plain CNN...")

start = time.time()

plain_history = plain_model.fit(
    x_train_norm,
    y_train,
    epochs=EPOCHS_CNN,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    verbose=1
)

plain_time = time.time() - start

# ------------------------------------------------------------
# 9. TRAIN RESIDUAL CNN
# ------------------------------------------------------------

print("\nTraining Residual CNN...")

start = time.time()

residual_history = residual_model.fit(
    x_train_norm,
    y_train,
    epochs=EPOCHS_CNN,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    verbose=1
)

residual_time = time.time() - start

# ------------------------------------------------------------
# 10. COMPARE PLAIN CNN AND RESIDUAL CNN
# ------------------------------------------------------------

plain_loss, plain_acc = plain_model.evaluate(
    x_test_norm, y_test, verbose=0
)

res_loss, res_acc = residual_model.evaluate(
    x_test_norm, y_test, verbose=0
)

print("\nCNN COMPARISON")
print("------------------------------")

print("Plain CNN Accuracy:", plain_acc)
print("Residual CNN Accuracy:", res_acc)

print("Plain CNN Training Time:", plain_time)
print("Residual CNN Training Time:", residual_time)

# Accuracy graph
plt.figure(figsize=(10, 5))

plt.plot(
    plain_history.history['val_accuracy'],
    label='Plain CNN'
)

plt.plot(
    residual_history.history['val_accuracy'],
    label='Residual CNN'
)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()

# Loss graph
plt.figure(figsize=(10, 5))

plt.plot(
    plain_history.history['val_loss'],
    label='Plain CNN'
)

plt.plot(
    residual_history.history['val_loss'],
    label='Residual CNN'
)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()


# ============================================================
# 11. PRETRAINED MOBILENETV2
# ============================================================

# MobileNetV2 expects larger images.
# Resize CIFAR-10 images from 32x32 to 96x96.

def prepare_mobilenet_data(images):

    images = tf.convert_to_tensor(images, dtype=tf.float32)

    images = tf.image.resize(
        images, (96, 96)
    )

    images = preprocess_input(images)

    return images


x_train_mobile = prepare_mobilenet_data(x_train)
x_test_mobile = prepare_mobilenet_data(x_test)

# ------------------------------------------------------------
# 12. DATA AUGMENTATION FOR MOBILENET
# ------------------------------------------------------------

mobile_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])

# ------------------------------------------------------------
# 13. LOAD PRETRAINED MOBILENETV2
# ------------------------------------------------------------

base_model = MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(96, 96, 3)
)

print("MobileNetV2 loaded successfully")

# ------------------------------------------------------------
# 14. FROZEN FEATURE EXTRACTION
# ------------------------------------------------------------

base_model.trainable = False

inputs = layers.Input(shape=(96, 96, 3))

x = mobile_augmentation(inputs)

# Training=False keeps pretrained BatchNorm layers
# in inference mode.
x = base_model(x, training=False)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dense(128, activation='relu')(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES, activation='softmax'
)(x)

frozen_model = models.Model(inputs, outputs)

frozen_model.compile(
    optimizer=tf.keras.optimizers.Adam(0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

frozen_model.summary()

print("\nTraining Frozen Feature Extraction Model...")

frozen_history = frozen_model.fit(
    x_train_mobile,
    y_train,
    epochs=EPOCHS_MOBILENET,
    batch_size=BATCH_SIZE,
    validation_split=0.2
)

frozen_loss, frozen_acc = frozen_model.evaluate(
    x_test_mobile, y_test, verbose=0
)

print("Frozen Feature Extraction Accuracy:", frozen_acc)


# ------------------------------------------------------------
# 15. PARTIAL FINE-TUNING
# ------------------------------------------------------------

# Unfreeze the last 20 layers of MobileNetV2.

base_model.trainable = True

for layer in base_model.layers[:-20]:
    layer.trainable = False

# Keep Batch Normalization layers frozen.
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

# Recompile after changing trainable layers.
frozen_model.compile(
    optimizer=tf.keras.optimizers.Adam(0.00001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nTraining Partial Fine-Tuning Model...")

fine_history = frozen_model.fit(
    x_train_mobile,
    y_train,
    epochs=EPOCHS_MOBILENET,
    batch_size=BATCH_SIZE,
    validation_split=0.2
)

fine_loss, fine_acc = frozen_model.evaluate(
    x_test_mobile, y_test, verbose=0
)

print("Partial Fine-Tuning Accuracy:", fine_acc)


# ============================================================
# 16. COMPARE MOBILENET MODELS
# ============================================================

print("\nMOBILENET COMPARISON")
print("------------------------------")

print("Frozen Feature Extraction:", frozen_acc)
print("Partial Fine-Tuning:", fine_acc)

plt.figure(figsize=(10, 5))

plt.plot(
    frozen_history.history['val_accuracy'],
    label='Frozen Feature Extraction'
)

plt.plot(
    fine_history.history['val_accuracy'],
    label='Partial Fine-Tuning'
)

plt.title("MobileNetV2 Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()


# ============================================================
# 17. FEATURE MAP VISUALIZATION
# ============================================================

# Display activations from the first convolution layer
# of the trained Residual CNN.

feature_layer = residual_model.get_layer(
    index=2
)

feature_model = models.Model(
    inputs=residual_model.input,
    outputs=feature_layer.output
)

sample_image = x_test_norm[0:1]

feature_maps = feature_model.predict(
    sample_image, verbose=0
)

plt.figure(figsize=(12, 8))

for i in range(min(16, feature_maps.shape[-1])):

    plt.subplot(4, 4, i + 1)

    plt.imshow(
        feature_maps[0, :, :, i],
        cmap='viridis'
    )

    plt.title(f"Feature Map {i+1}")
    plt.axis('off')

plt.suptitle("CNN Feature Map Visualization")
plt.tight_layout()
plt.show()


# ============================================================
# 18. GRAD-CAM VISUALIZATION
# ============================================================

def make_gradcam_heatmap(
    image,
    model,
    last_conv_layer_name
):

    grad_model = tf.keras.models.Model(
        model.inputs,
        [
            model.get_layer(
                last_conv_layer_name
            ).output,
            model.output
        ]
    )

    with tf.GradientTape() as tape:

        conv_outputs, predictions = grad_model(
            image, training=False
        )

        predicted_class = tf.argmax(
            predictions[0]
        )

        class_score = predictions[:, predicted_class]

    grads = tape.gradient(
        class_score,
        conv_outputs
    )

    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0, 1, 2)
    )

    conv_outputs = conv_outputs[0]

    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    heatmap = tf.maximum(heatmap, 0)

    heatmap = heatmap / (
        tf.reduce_max(heatmap) + 1e-8
    )

    return heatmap.numpy(), int(predicted_class.numpy())


# Find the final convolution layer of Residual CNN.
last_conv_layer = None

for layer in reversed(residual_model.layers):

    if isinstance(layer, layers.Conv2D):
        last_conv_layer = layer.name
        break

print("Last convolution layer:", last_conv_layer)

# Select a test image
image_index = 0

image_for_gradcam = x_test_norm[
    image_index:image_index+1
]

heatmap, predicted_class = make_gradcam_heatmap(
    image_for_gradcam,
    residual_model,
    last_conv_layer
)

# Resize heatmap to original image dimensions
heatmap_resized = tf.image.resize(
    heatmap[..., np.newaxis],
    (32, 32)
).numpy().squeeze()

original_image = x_test[image_index]

# Display original image and Grad-CAM
plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)

plt.imshow(original_image)

plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)

plt.imshow(heatmap_resized, cmap='jet')

plt.title("Grad-CAM Heatmap")
plt.axis('off')

plt.subplot(1, 3, 3)

plt.imshow(original_image)

plt.imshow(
    heatmap_resized,
    cmap='jet',
    alpha=0.4
)

plt.title("Grad-CAM Overlay")
plt.axis('off')

plt.suptitle(
    "Predicted Class: " + class_names[predicted_class]
)

plt.tight_layout()
plt.show()


# ============================================================
# 19. FINAL COMPARISON
# ============================================================

print("\n========== FINAL RESULTS ==========")

print(f"Plain CNN Accuracy: {plain_acc*100:.2f}%")
print(f"Residual CNN Accuracy: {res_acc*100:.2f}%")

print(f"Frozen MobileNetV2 Accuracy: {frozen_acc*100:.2f}%")
print(f"Fine-Tuned MobileNetV2 Accuracy: {fine_acc*100:.2f}%")

print("\nProgram Completed Successfully!")
```

```python

!pip install -q datasets huggingface_hub

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from datasets import load_dataset
from huggingface_hub import HfApi, hf_hub_download
from tensorflow.keras import layers, models
from PIL import Image
import os

tf.keras.utils.set_random_seed(42)

print("TensorFlow Version:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))

print("\n1. HUGGING FACE HUB")

api = HfApi()

dataset_results = list(api.list_datasets(search="cifar10", limit=5))

print("Available CIFAR-10 datasets:")

for dataset in dataset_results:
    print(dataset.id)

dataset_id = "uoft-cs/cifar10"

print("\nSelected Dataset:", dataset_id)

try:
    dataset_card_path = hf_hub_download(
        repo_id=dataset_id,
        filename="README.md",
        repo_type="dataset"
    )

    with open(dataset_card_path, "r", encoding="utf-8") as file:
        print("\nDataset Model Card:")
        print(file.read()[:2000])

except Exception as e:
    print("Dataset card could not be retrieved:", e)


print("\n2. HUGGING FACE MODEL CARD")

model_id = "google/vit-base-patch16-224"

try:
    model_info = api.model_info(model_id)

    print("Model ID:", model_info.id)
    print("Model Pipeline:", model_info.pipeline_tag)
    print("Model Downloads:", model_info.downloads)
    print("Model Likes:", model_info.likes)

    model_card_path = hf_hub_download(
        repo_id=model_id,
        filename="README.md"
    )

    with open(model_card_path, "r", encoding="utf-8") as file:
        print("\nModel Card Content:")
        print(file.read()[:2000])

except Exception as e:
    print("Model card retrieval error:", e)


print("\n3. LOAD CIFAR-10 FROM HUGGING FACE")

dataset = load_dataset(dataset_id)

print(dataset)
print("Training features:", dataset["train"].features)
print("Testing features:", dataset["test"].features)

train_data = dataset["train"].select(range(10000))
test_data = dataset["test"].select(range(2000))

image_column = next(
    col for col in train_data.column_names
    if col in ["img", "image"]
)

def convert_images(data):
    images = np.stack([
        np.asarray(image.convert("RGB"))
        for image in data[image_column]
    ])

    labels = np.array(data["label"])

    return images.astype("float32") / 255.0, labels

x_train, y_train = convert_images(train_data)
x_test, y_test = convert_images(test_data)

print("Training Shape:", x_train.shape)
print("Testing Shape:", x_test.shape)

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[y_train[i]])
    plt.axis("off")

plt.tight_layout()
plt.show()


print("\n4. NUMPY CONVOLUTION IMPLEMENTATION")

def convolution(image, kernel, stride=1, padding="valid"):

    h, w, channels = image.shape
    kh, kw, _, filters = kernel.shape

    if padding.lower() == "same":
        out_h = int(np.ceil(h / stride))
        out_w = int(np.ceil(w / stride))

        pad_h = max((out_h - 1) * stride + kh - h, 0)
        pad_w = max((out_w - 1) * stride + kw - w, 0)

        top = pad_h // 2
        bottom = pad_h - top
        left = pad_w // 2
        right = pad_w - left

        image = np.pad(
            image,
            ((top, bottom), (left, right), (0, 0))
        )

    elif padding.lower() == "valid":
        out_h = (h - kh) // stride + 1
        out_w = (w - kw) // stride + 1

    else:
        raise ValueError("Padding must be 'same' or 'valid'")

    output = np.zeros((out_h, out_w, filters))

    for i in range(out_h):
        for j in range(out_w):

            region = image[
                i * stride:i * stride + kh,
                j * stride:j * stride + kw,
                :
            ]

            for f in range(filters):
                output[i, j, f] = np.sum(
                    region * kernel[:, :, :, f]
                )

    return output


sample_image = x_train[0]

kernel = np.random.randn(3, 3, 3, 2)

conv_output = convolution(
    sample_image,
    kernel,
    stride=1,
    padding="valid"
)

print("Original Image Shape:", sample_image.shape)
print("Kernel Shape:", kernel.shape)
print("Convolution Output Shape:", conv_output.shape)


print("\n5. MAX POOLING IMPLEMENTATION")

def max_pooling(image, pool_size=2, stride=2):

    h, w, channels = image.shape

    out_h = (h - pool_size) // stride + 1
    out_w = (w - pool_size) // stride + 1

    output = np.zeros((out_h, out_w, channels))

    for i in range(out_h):
        for j in range(out_w):

            region = image[
                i * stride:i * stride + pool_size,
                j * stride:j * stride + pool_size,
                :
            ]

            output[i, j, :] = np.max(
                region,
                axis=(0, 1)
            )

    return output


pool_output = max_pooling(
    conv_output,
    pool_size=2,
    stride=2
)

print("Before Pooling:", conv_output.shape)
print("After Pooling:", pool_output.shape)


print("\n6. KERNEL SIZE, STRIDE AND PADDING")

kernel_results = []

for kernel_size in [3, 5]:
    kernel = np.random.randn(
        kernel_size,
        kernel_size,
        3,
        4
    )

    for stride in [1, 2]:
        for padding in ["valid", "same"]:

            output = convolution(
                sample_image,
                kernel,
                stride=stride,
                padding=padding
            )

            kernel_results.append({
                "Kernel Size": kernel_size,
                "Stride": stride,
                "Padding": padding,
                "Output Shape": output.shape
            })

kernel_df = pd.DataFrame(kernel_results)

print(kernel_df.to_string(index=False))


print("\n7. CNN MODEL FOR CIFAR-10")

cnn_model = models.Sequential([

    layers.Input(shape=(32, 32, 3)),

    layers.Conv2D(
        32,
        kernel_size=3,
        strides=1,
        padding="same",
        activation="relu"
    ),

    layers.MaxPooling2D(pool_size=2),

    layers.Conv2D(
        64,
        kernel_size=3,
        strides=1,
        padding="same",
        activation="relu"
    ),

    layers.MaxPooling2D(pool_size=2),

    layers.Conv2D(
        128,
        kernel_size=3,
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(pool_size=2),

    layers.Flatten(),

    layers.Dense(128, activation="relu"),

    layers.Dropout(0.3),

    layers.Dense(10, activation="softmax")
])

cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn_model.summary()


print("\n8. CNN TRAINING")

history = cnn_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)


print("\n9. TRAINING AND VALIDATION GRAPHS")

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.title("CNN Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("CNN Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


print("\n10. FINAL TEST EVALUATION")

test_loss, test_accuracy = cnn_model.evaluate(
    x_test,
    y_test,
    verbose=1
)

print("Test Loss:", round(test_loss, 4))
print("Test Accuracy:", round(test_accuracy, 4))


print("\n11. SAMPLE PREDICTIONS")

predictions = cnn_model.predict(x_test[:10])

predicted_classes = np.argmax(predictions, axis=1)

plt.figure(figsize=(12, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(x_test[i])

    plt.title(
        "Actual: " + class_names[y_test[i]] +
        "\nPredicted: " + class_names[predicted_classes[i]]
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


print("\n12. SAVE RESULTS")

results = pd.DataFrame({
    "Metric": [
        "Test Loss",
        "Test Accuracy",
        "Training Samples",
        "Testing Samples"
    ],
    "Value": [
        test_loss,
        test_accuracy,
        len(x_train),
        len(x_test)
    ]
})

print(results)

results.to_csv("cifar10_cnn_results.csv", index=False)

cnn_model.save("cifar10_cnn_model.keras")

print("\nExperiment execution finished.")
```

```python

# ============================================================
# DEEP LEARNING - SELF LEARNING LAB PROGRAM 1
# Fashion-MNIST | Complete Google Colab Implementation
# ============================================================

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers.schedules import ExponentialDecay

import warnings
warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

SEED = 42
EPOCHS = 8
BATCH_SIZE = 128

tf.keras.utils.set_random_seed(SEED)

print("=" * 65)
print("DEEP LEARNING - SELF LEARNING LAB PROGRAM 1")
print("=" * 65)

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))

# ------------------------------------------------------------
# 1. LOAD AND PREPROCESS FASHION-MNIST
# ------------------------------------------------------------

print("\n1. LOADING FASHION-MNIST DATASET")

(x_full, y_full), (x_test_full, y_test_full) = (
    tf.keras.datasets.fashion_mnist.load_data()
)

# Select subset
x_train = x_full[:10000].astype("float32") / 255.0
y_train = y_full[:10000]

x_test = x_test_full[:2000].astype("float32") / 255.0
y_test = y_test_full[:2000]

# Training and validation split
x_val = x_train[8000:10000]
y_val = y_train[8000:10000]

x_train = x_train[:8000]
y_train = y_train[:8000]

# Add channel dimension
x_train = x_train[..., np.newaxis]
x_val = x_val[..., np.newaxis]
x_test = x_test[..., np.newaxis]

class_names = [
    "T-shirt", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

print("Training samples:", len(x_train))
print("Validation samples:", len(x_val))
print("Testing samples:", len(x_test))

# Visualize sample images
plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i].squeeze(), cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.axis("off")

plt.suptitle("Fashion-MNIST Sample Images")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 2. MLP MODEL FUNCTION
# ------------------------------------------------------------

def build_mlp(
    optimizer,
    dropout_rate=0.0,
    l2_rate=0.0,
    augmentation=False
):

    inputs = layers.Input(shape=(28, 28, 1))
    x = inputs

    # Data augmentation
    if augmentation:
        x = layers.RandomTranslation(0.1, 0.1)(x)
        x = layers.RandomRotation(0.05)(x)
        x = layers.RandomZoom(0.1)(x)

    x = layers.Flatten()(x)

    reg = (
        regularizers.l2(l2_rate)
        if l2_rate > 0 else None
    )

    x = layers.Dense(
        256,
        activation="relu",
        kernel_regularizer=reg
    )(x)

    if dropout_rate > 0:
        x = layers.Dropout(dropout_rate)(x)

    x = layers.Dense(
        128,
        activation="relu",
        kernel_regularizer=reg
    )(x)

    if dropout_rate > 0:
        x = layers.Dropout(dropout_rate)(x)

    outputs = layers.Dense(10, activation="softmax")(x)

    model = models.Model(inputs, outputs)

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ------------------------------------------------------------
# 3. BATCH, STOCHASTIC AND MINI-BATCH GRADIENT DESCENT
# ------------------------------------------------------------

print("\n2. GRADIENT DESCENT COMPARISON")

demo_x = x_train[:2000]
demo_y = y_train[:2000]

batch_sizes = {
    "Batch GD": len(demo_x),
    "Stochastic GD": 1,
    "Mini-Batch GD": 128
}

# Evaluate every model on the same fixed dataset
class FixedEvaluation(tf.keras.callbacks.Callback):

    def __init__(self, x, y):
        super().__init__()
        self.x = x
        self.y = y
        self.loss_values = []
        self.accuracy_values = []

    def on_epoch_end(self, epoch, logs=None):

        loss, accuracy = self.model.evaluate(
            self.x, self.y, verbose=0
        )

        self.loss_values.append(loss)
        self.accuracy_values.append(accuracy)


gd_results = {}

for name, batch_size in batch_sizes.items():

    print("\nTraining:", name)

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(SEED)

    model = build_mlp(
        tf.keras.optimizers.SGD(learning_rate=0.01)
    )

    evaluator = FixedEvaluation(demo_x, demo_y)

    model.fit(
        demo_x,
        demo_y,
        epochs=3,
        batch_size=batch_size,
        callbacks=[evaluator],
        verbose=1
    )

    gd_results[name] = {
        "loss": evaluator.loss_values,
        "accuracy": evaluator.accuracy_values
    }

    updates = int(np.ceil(len(demo_x) / batch_size))

    print("Updates per epoch:", updates)


# Plot gradient descent comparison
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

for name, result in gd_results.items():
    plt.plot(result["loss"], marker="o", label=name)

plt.title("Gradient Descent - Evaluation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)

for name, result in gd_results.items():
    plt.plot(result["accuracy"], marker="o", label=name)

plt.title("Gradient Descent - Evaluation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 4. OPTIMIZER AND LEARNING RATE COMPARISON
# ------------------------------------------------------------

print("\n3. OPTIMIZER COMPARISON")

optimizer_results = {}

optimizer_names = [
    "SGD Momentum",
    "RMSprop",
    "Adam"
]

steps_per_epoch = int(np.ceil(len(x_train) / BATCH_SIZE))

def create_optimizer(name, scheduled=False):

    # Different standard starting rates for each optimizer
    initial_lr = (
        0.01 if name == "SGD Momentum"
        else 0.001
    )

    if scheduled:
        lr = ExponentialDecay(
            initial_learning_rate=initial_lr,
            decay_steps=steps_per_epoch * 3,
            decay_rate=0.5,
            staircase=False
        )
    else:
        lr = initial_lr

    if name == "SGD Momentum":
        return tf.keras.optimizers.SGD(
            learning_rate=lr,
            momentum=0.9
        )

    elif name == "RMSprop":
        return tf.keras.optimizers.RMSprop(
            learning_rate=lr
        )

    elif name == "Adam":
        return tf.keras.optimizers.Adam(
            learning_rate=lr
        )


for optimizer_name in optimizer_names:

    for scheduled in [False, True]:

        rate_type = (
            "Scheduled" if scheduled else "Constant"
        )

        experiment_name = optimizer_name + " - " + rate_type

        print("\nTraining:", experiment_name)

        tf.keras.backend.clear_session()
        tf.keras.utils.set_random_seed(SEED)

        optimizer = create_optimizer(
            optimizer_name,
            scheduled
        )

        model = build_mlp(optimizer)

        history = model.fit(
            x_train,
            y_train,
            epochs=EPOCHS,
            batch_size=BATCH_SIZE,
            validation_data=(x_val, y_val),
            verbose=1
        )

        optimizer_results[experiment_name] = history.history


# Optimizer results table
optimizer_summary = []

for name, history in optimizer_results.items():

    optimizer_summary.append({
        "Optimizer": name,
        "Training Accuracy": history["accuracy"][-1],
        "Validation Accuracy": history["val_accuracy"][-1],
        "Training Loss": history["loss"][-1],
        "Validation Loss": history["val_loss"][-1]
    })

optimizer_df = pd.DataFrame(optimizer_summary)

print("\nOPTIMIZER RESULTS")
print(optimizer_df.round(4).to_string(index=False))


# Optimizer graphs
plt.figure(figsize=(12, 5))

for name, history in optimizer_results.items():
    plt.plot(
        history["val_accuracy"],
        marker="o",
        label=name
    )

plt.title("Optimizer Comparison - Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(12, 5))

for name, history in optimizer_results.items():
    plt.plot(
        history["val_loss"],
        marker="o",
        label=name
    )

plt.title("Optimizer Comparison - Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid()
plt.show()


# ------------------------------------------------------------
# 5. OVERFITTING INVESTIGATION
# ------------------------------------------------------------

print("\n4. OVERFITTING INVESTIGATION")

tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

overfit_model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Flatten(),
    layers.Dense(512, activation="relu"),
    layers.Dense(512, activation="relu"),
    layers.Dense(512, activation="relu"),
    layers.Dense(10, activation="softmax")
])

overfit_model.compile(
    optimizer=tf.keras.optimizers.Adam(0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

overfit_history = overfit_model.fit(
    x_train,
    y_train,
    epochs=15,
    batch_size=128,
    validation_data=(x_val, y_val),
    verbose=1
)

print(
    "Final Training Accuracy:",
    round(overfit_history.history["accuracy"][-1], 4)
)

print(
    "Final Validation Accuracy:",
    round(overfit_history.history["val_accuracy"][-1], 4)
)

# Overfitting graphs
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(overfit_history.history["accuracy"], label="Training")
plt.plot(overfit_history.history["val_accuracy"], label="Validation")
plt.title("Overfitting - Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(overfit_history.history["loss"], label="Training")
plt.plot(overfit_history.history["val_loss"], label="Validation")
plt.title("Overfitting - Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. DROPOUT, L2 AND DATA AUGMENTATION
# ------------------------------------------------------------

print("\n5. REGULARIZATION COMPARISON")

regularization_configs = {

    "Baseline": {
        "dropout_rate": 0.0,
        "l2_rate": 0.0,
        "augmentation": False
    },

    "Dropout": {
        "dropout_rate": 0.3,
        "l2_rate": 0.0,
        "augmentation": False
    },

    "L2 Regularization": {
        "dropout_rate": 0.0,
        "l2_rate": 0.001,
        "augmentation": False
    },

    "Data Augmentation": {
        "dropout_rate": 0.0,
        "l2_rate": 0.0,
        "augmentation": True
    },

    "Combined": {
        "dropout_rate": 0.3,
        "l2_rate": 0.001,
        "augmentation": True
    }
}

regularization_results = {}
best_regularized_model = None
best_regularized_accuracy = -1

for name, config in regularization_configs.items():

    print("\nTraining:", name)

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(SEED)

    model = build_mlp(
        optimizer=tf.keras.optimizers.Adam(0.001),
        **config
    )

    history = model.fit(
        x_train,
        y_train,
        epochs=10,
        batch_size=BATCH_SIZE,
        validation_data=(x_val, y_val),
        verbose=1
    )

    regularization_results[name] = history.history

    best_accuracy = max(history.history["val_accuracy"])

    if best_accuracy > best_regularized_accuracy:
        best_regularized_accuracy = best_accuracy
        best_regularized_model = model


# Regularization result table
regularization_summary = []

for name, history in regularization_results.items():

    regularization_summary.append({
        "Technique": name,
        "Training Accuracy": history["accuracy"][-1],
        "Validation Accuracy": history["val_accuracy"][-1],
        "Training Loss": history["loss"][-1],
        "Validation Loss": history["val_loss"][-1]
    })

regularization_df = pd.DataFrame(regularization_summary)

print("\nREGULARIZATION RESULTS")
print(regularization_df.round(4).to_string(index=False))


# Regularization graph
plt.figure(figsize=(12, 6))

for name, history in regularization_results.items():
    plt.plot(
        history["val_accuracy"],
        marker="o",
        label=name
    )

plt.title("Regularization Techniques - Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid()
plt.show()


# ------------------------------------------------------------
# 7. EARLY STOPPING
# ------------------------------------------------------------

print("\n6. EARLY STOPPING")

tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

early_model = build_mlp(
    optimizer=tf.keras.optimizers.Adam(0.001),
    dropout_rate=0.3,
    l2_rate=0.001
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True,
    verbose=1
)

early_history = early_model.fit(
    x_train,
    y_train,
    epochs=30,
    batch_size=BATCH_SIZE,
    validation_data=(x_val, y_val),
    callbacks=[early_stopping],
    verbose=1
)

print(
    "Epochs executed:",
    len(early_history.history["loss"])
)

plt.figure(figsize=(10, 5))

plt.plot(early_history.history["loss"], label="Training Loss")
plt.plot(early_history.history["val_loss"], label="Validation Loss")

plt.title("Early Stopping")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()


# ------------------------------------------------------------
# 8. PLAIN MLP AND RESIDUAL MLP
# ------------------------------------------------------------

print("\n7. PLAIN MLP VS RESIDUAL MLP")

def build_plain_model():

    inputs = layers.Input(shape=(28, 28, 1))
    x = layers.Flatten()(inputs)

    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dense(256, activation="relu")(x)

    outputs = layers.Dense(10, activation="softmax")(x)

    model = models.Model(inputs, outputs)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(0.001),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


def build_residual_model():

    inputs = layers.Input(shape=(28, 28, 1))
    x = layers.Flatten()(inputs)

    # Initial layer
    x = layers.Dense(256, activation="relu")(x)

    # Residual block
    shortcut = x

    y = layers.Dense(256, activation="relu")(x)
    y = layers.Dense(256)(y)

    x = layers.Add()([shortcut, y])
    x = layers.ReLU()(x)

    outputs = layers.Dense(10, activation="softmax")(x)

    model = models.Model(inputs, outputs)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(0.001),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# Train Plain MLP
tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

plain_model = build_plain_model()

plain_history = plain_model.fit(
    x_train,
    y_train,
    epochs=15,
    batch_size=BATCH_SIZE,
    validation_data=(x_val, y_val),
    verbose=1
)

# Train Residual MLP
tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

residual_model = build_residual_model()

residual_history = residual_model.fit(
    x_train,
    y_train,
    epochs=15,
    batch_size=BATCH_SIZE,
    validation_data=(x_val, y_val),
    verbose=1
)


# Architecture comparison graphs
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    plain_history.history["val_accuracy"],
    label="Plain MLP"
)

plt.plot(
    residual_history.history["val_accuracy"],
    label="Residual MLP"
)

plt.title("Plain vs Residual - Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)

plt.plot(
    plain_history.history["val_loss"],
    label="Plain MLP"
)

plt.plot(
    residual_history.history["val_loss"],
    label="Residual MLP"
)

plt.title("Plain vs Residual - Loss")
plt.xlabel("Epochs")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. FINAL MODEL EVALUATION
# ------------------------------------------------------------

print("\n8. FINAL TEST EVALUATION")

models_to_evaluate = {
    "Plain MLP": plain_model,
    "Residual MLP": residual_model,
    "Early Stopping MLP": early_model,
    "Best Regularized MLP": best_regularized_model
}

final_results = []

for name, model in models_to_evaluate.items():

    test_loss, test_accuracy = model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    final_results.append({
        "Model": name,
        "Test Loss": round(test_loss, 4),
        "Test Accuracy": round(test_accuracy, 4)
    })

final_df = pd.DataFrame(final_results)

print("\nFINAL MODEL PERFORMANCE")
print(final_df.to_string(index=False))


# ------------------------------------------------------------
# 10. SAVE EXPERIMENTAL RESULTS
# ------------------------------------------------------------

optimizer_df.to_csv("optimizer_results.csv", index=False)

regularization_df.to_csv(
    "regularization_results.csv",
    index=False
)

final_df.to_csv(
    "final_model_results.csv",
    index=False
)

# Save all result tables together
with pd.ExcelWriter("DL_Lab_Program_1_Results.xlsx") as writer:

    optimizer_df.to_excel(
        writer, sheet_name="Optimizers", index=False
    )

    regularization_df.to_excel(
        writer, sheet_name="Regularization", index=False
    )

    final_df.to_excel(
        writer, sheet_name="Final Evaluation", index=False
    )

print("\n" + "=" * 65)
print("ALL EXPERIMENTS COMPLETED SUCCESSFULLY")
print("=" * 65)

print("\nRESULT FILES CREATED:")
print("1. optimizer_results.csv")
print("2. regularization_results.csv")
print("3. final_model_results.csv")
print("4. DL_Lab_Program_1_Results.xlsx")

print("\nLAB CONCLUSION:")
print("1. Batch, Stochastic and Mini-Batch GD were demonstrated.")
print("2. Momentum, RMSprop and Adam were compared.")
print("3. Constant and scheduled learning rates were evaluated.")
print("4. Overfitting was investigated.")
print("5. Dropout and L2 regularization were implemented.")
print("6. Data augmentation was applied during training.")
print("7. Early stopping was implemented.")
print("8. Plain and Residual MLP architectures were compared.")
print("9. Final models were evaluated on unseen test data.")

print("\nDownload your Excel results from the Colab Files panel.")
```
